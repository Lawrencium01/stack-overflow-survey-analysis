```python
def split_multicolumn(col_series):
    result_df = col_series.to_frame()
    options = []

    # Iterate over the column
    for idx, value in col_series[col_series.notnull()].items():

        # Break each value into list of options
        for option in value.split(';'):

            # Add the option as a column to result
            if option not in result_df.columns:
                options.append(option)
                result_df[option] = False

            # Mark the value in the option column as True
            result_df.at[idx, option] = True

    return result_df[options]
```

```python
dev_type_df = split_multicolumn(survery_df['DevType'])
```

```python
dev_type_df
```

```python
dev_type_totals = dev_type_df.sum().sort_values(ascending = False)
dev_type_totals
```

## Asking and Answering Questions

```python
survery_df['LanguageWorkedWith'].value_counts().head()
```

```python
languages_worked_df = split_multicolumn(survery_df['LanguageWorkedWith'])
```

```python
languages_worked_df
```

```python
languages_worked_percentages = languages_worked_df.mean().sort_values(ascending = False) * 100
languages_worked_percentages
```

```python
plt.figure(figsize = (12,12))
sns.barplot(y = languages_worked_percentages.index, x = languages_worked_percentages, color = 'r')
plt.title('Languages used in the past year')
plt.xlabel('Count');
```

```python
languages_interested_df = split_multicolumn(survery_df['LanguageDesireNextYear'])
languages_interested_percentages = languages_interested_df.mean().sort_values(ascending = False)
languages_interested_percentages
```

```python
plt.figure(figsize = (12,12))
sns.barplot(y = languages_interested_percentages.index, x = languages_interested_percentages)
plt.title('Languages people are  interested in learing over the next year')
plt.xlabel('count');
```
