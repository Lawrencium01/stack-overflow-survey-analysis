### what is the total number of reported cases and reported deaths related to covid-19 in italy

```python
total_cases = covid_df['new_cases'].sum()
total_cases
```

```python
total_deaths = covid_df['new_deaths'].sum()
total_deaths
```

```python
print(f'The total number of reported cases is {total_cases} and the total number of reported death is {total_deaths}')
```

### What is the Overall Death Rate

```python
death_rate = total_deaths / total_cases
death_rate
```

```python
print(f'The Overall reported death rate in italy is {death_rate: {'0.2f'}} ')
```

### What is the overall number of tests conducted? A total of 935310 tests were conducted before daily test numbers were being reported

```python
initial_tests = 935310
total_tests = initial_tests + covid_df['new_tests'].sum()
total_tests
```

### What fraction of test returned a positive result

```python
positive_rate = total_cases / total_tests
positive_rate
```

```python
print(f'{positive_rate: {'0.2f'}} in Italy led to a positive diagnosis')
```

## Merging data from multiple sources

```python
locations_df = pd.read_csv(r"C:\Users\HP\Documents\Learning Data Analytics\Python\FreeCodeCamp\Covid\location.csv")
```

```python
locations_df
```

```python
locations_df[locations_df['location'] == 'Italy']
```

```python
covid_df['location'] = 'Italy'
```

```python
covid_df
```

```python
merged_df = covid_df.merge(locations_df, on = 'location')
```

```python
merged_df
```

```python
merged_df['cases_per_million'] = merged_df['total_cases'] * 1e6 / merged_df['population']
```

```python
merged_df['deaths_per_million'] = merged_df['total_deaths'] * 1e6 / merged_df['population']
```

```python
merged_df['tests_per_million'] = merged_df['total_tests'] * 1e6 / merged_df['population']
```

```python
merged_df
```

```python
result_df = merged_df[['date','new_cases','total_cases','new_deaths','total_deaths','new_tests','total_tests','cases_per_million','deaths_per_million','tests_per_million']]
result_df
```
